## Example

```python
distances = torch.tensor([
    [1.0], [1.5], [2.0], [2.5], [3.0], [3.5], [4.0], [4.5], [5.0], [5.5],
    [6.0], [6.5], [7.0], [7.5], [8.0], [8.5], [9.0], [9.5], [10.0], [10.5],
    [11.0], [11.5], [12.0], [12.5], [13.0], [13.5], [14.0], [14.5], [15.0], [15.5],
    [16.0], [16.5], [17.0], [17.5], [18.0], [18.5], [19.0], [19.5], [20.0]
], dtype=torch.float32)

times = torch.tensor([
    [6.96], [9.67], [12.11], [14.56], [16.77], [21.7], [26.52], [32.47], [37.15], [42.35],
    [46.1], [52.98], [57.76], [61.29], [66.15], [67.63], [69.45], [71.57], [72.8], [73.88],
    [76.34], [76.38], [78.34], [80.07], [81.86], [84.45], [83.98], [86.55], [88.33], [86.83],
    [89.24], [88.11], [88.16], [91.77], [92.27], [92.13], [90.73], [90.39], [92.98]
], dtype=torch.float32)
```

```python
distances.dtype, distances.ndim, distances.size()
```

```python
## Plotting the data points


plt.scatter(distances, times)
```

```python
distances_mean,distances_std,times_mean, times_std = distances.mean(), distances.std(), times.mean(),times.std()
distances_norm = (distances - distances_mean)/distances_std
times_norm = (times - times_mean)/times_std

plt.scatter(distances_norm, times_norm)
```

```python
torch.manual_seed(42)

# model = nn.Sequential(
#     nn.Linear(1,1)
# )
model = nn.Sequential(
    nn.Linear(1,3),
    nn.ReLU(), # ReLU: max(0,z)
    nn.Linear(3,1)
)
```

### Loss Function and Optimizer
Loss Function: Measures how wrong the neural network's predictions are <br>
Optimizer: Updates the weights and biases to reduce that error.

```python
loss_function = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
# Training loop
model.train()
for epoch in range(3000):
      # Reset the optimizer's gradients
    optimizer.zero_grad()
    # Make predictions (forward pass)
    outputs = model(distances_norm)
    # Calculate the loss
    loss = loss_function(outputs, times_norm)
    # Calculate adjustments (backward pass)
    loss.backward()
    # Update the model's parameters
    optimizer.step()

    if (epoch + 1) % 50 == 0:
      print(loss)

print("\nTraining Complete.")
print(f"\nFinal Loss: {loss.item()}")
```

```python
model.eval()

with torch.no_grad():
  predicted_norm = model(distances_norm)
  predicted_times = (predicted_norm * times_std) + times_mean
  plt.figure(figsize=(8, 6))
  plt.plot(distances.numpy(), times.numpy(), color='orange', marker='o', linestyle='none', label='Actual Data')
  plt.plot(distances.numpy(), predicted_times.numpy(), color='green', label='Non-Linear Model Predictions')
  plt.title(' Model Fit ')
  plt.xlabel('Distance (miles)')
  plt.ylabel('Time (minutes)')
  plt.legend()
  plt.grid(True)
  plt.show()
```

## Why clear the gradients (optimizer.zero_grad())??

```python
w = torch.tensor(2.0, requires_grad=True)

loss = w ** 2
loss.backward()
print(w.grad)

loss = w ** 2
loss.backward()
print(w.grad)

loss = w ** 2
loss.backward()
print(w.grad)
```

## Autograd

```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# Forward pass
y = (w * x + b) ** 2

# Backward pass
y.backward()

print(w.grad)
print(b.grad)
```
