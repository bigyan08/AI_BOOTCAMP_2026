```python
import numpy as np

# Input feature: hours studied
X = np.array([1, 2, 3, 4, 5], dtype=float)

# Actual target: exam marks
y = np.array([20, 30, 40, 50, 60], dtype=float)

print("Hours:", X)
print("Marks:", y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.title("Study Hours vs Exam Marks")
plt.grid(True)
plt.show()
```

```python
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000
n = len(X)
```

```python

for epoch in range(epochs):

    # 1. Make predictions
    y_pred = w * X + b

    # 2. Calculate gradients
    dw = (2 / n) * np.sum((y_pred - y) * X)
    db = (2 / n) * np.sum(y_pred - y)

    # 3. Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # 4. Show progress occasionally
    if epoch % 100 == 0:
        loss = np.mean((y_pred - y) ** 2)
        print(
            f"Epoch {epoch}: "
            f"Loss={loss:.4f}, w={w:.4f}, b={b:.4f}"
        )

print("Final weight:", w)
print("Final bias:", b)
```

```python

new_hours = 6

predicted_marks = w * new_hours + b

print("Hours studied:", new_hours)
print("Predicted marks:", predicted_marks)
```

```python

plt.scatter(X, y, label="Actual data")

plt.plot(
    X,
    w * X + b,
    color="red",
    label="Learned regression line"
)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.title("Linear Regression From Scratch")
plt.legend()
plt.grid(True)
plt.show()
```
